```python

# Self-Reflex Agent: Weather Prediction System

class WeatherAgent:

    def __init__(self):
        self.score = 0
        self.memory = []

    def predict(self, temperature, humidity):

        if humidity > 70 and temperature < 25:
            prediction = "Rainy"

        elif humidity < 40:
            prediction = "Sunny"

        else:
            prediction = "Cloudy"

        return prediction

    def reflect(self, prediction, actual_weather):

        print("Predicted Weather:", prediction)
        print("Actual Weather:", actual_weather)

        if prediction == actual_weather:
            feedback = "Correct prediction!"
            self.score += 10

        else:
            feedback = "Wrong prediction. The agent needs improvement."

        self.memory.append(feedback)

        print("Reflection:", feedback)
        print("Performance Score:", self.score)
        print("-" * 40)


# Create agent
agent = WeatherAgent()

# Test cases
weather_data = [
    (20, 80, "Rainy"),
    (30, 30, "Sunny"),
    (25, 60, "Rainy"),
    (22, 50, "Cloudy")
]

# Run agent
for temperature, humidity, actual_weather in weather_data:

    prediction = agent.predict(temperature, humidity)

    agent.reflect(prediction, actual_weather)


# Final report
print("\nFINAL REPORT")
print("Total Score:", agent.score)

print("\nAgent Memory:")
for item in agent.memory:
    print(item)
```
